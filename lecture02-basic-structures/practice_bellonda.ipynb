```python
# 1. Create the list of ratings
hotel_ratings = [3.5, 4.2, 2.8, 5.0]

# 2. Grab the first rating (Remember: Python starts counting at 0!)
first_hotel = hotel_ratings[0]

# Let's see what we got
print("The first hotel rating is:", first_hotel)
```

```python
hotel_ratings[2] = 3.8
print("Fixed ratings:", hotel_ratings)
```

```python
hotel_ratings.append(4.9)
print(hotel_ratings)
```

```python
# Our ratings: [3.5, 4.2, 3.8, 5.0, 4.9]
# Corresponding prices: 80, 120, 100, 300, 250

# 1. Create the price list
hotel_prices = [ 80, 120, 100, 300, 250 ]

# 2. Grab the last price (The luxury one)
last_price = hotel_prices[-1]

# 3. Grab the last rating
last_rating = hotel_ratings[-1]

# 4. Calculate Value (Rating divided by Price)
value_score = last_rating / last_price

print("Value Score for Luxury Hotel:", value_score)
```

```python
# Create a dictionary for the luxury hotel
# Syntax: {'Key': Value, 'Key': Value}

luxury_profile = {'rating': 4.9, 'price': 250}

# Let's access the data by name (Key) instead of index
print("The price is:", luxury_profile['price'])
```

```python
# 1. Import the library
import pandas as pd

# 2. Prepare the data payload
# We map a "String Name" to the "List Variable"
data_payload = {
    'Rating': hotel_ratings,
    'Price': hotel_prices
}

# 3. Construct the DataFrame
df = pd.DataFrame(data_payload)

# Let's verify the structure
print(df)
```

```python
df['value'] = df['Rating'] / df['Price']
print(df)
```

```python
leaderboard = df.sort_values(by='value',ascending = False)
print(leaderboard)
```

```python
mask = df['Price'] < 150
affordable_hotels = df[mask]
print(affordable_hotels)
```

# 📜 The Grimoire: Syntax Card (Session 1)

| **Concept** | **The Syntax** | **Bellonda's Translation** |
| :--- | :--- | :--- |
| **Create List** | `x = [1, 2, 3]` | A conga line of data. Order matters. |
| **Access Item** | `x[0]` or `x[-1]` | Grab the first guy (`0`) or the last guy (`-1`). |
| **Append Item** | `x.append(4)` | Tack a new carriage onto the end of the train. |
| **Dictionary** | `{'key': value}` | A labeled profile. Order doesn't matter, names do. |
| **Create DataFrame** | `pd.DataFrame(data)` | Turning raw lists/dicts into a professional table. |
| **Vector Math** | `df['A'] / df['B']` | Doing math on the *entire* column at once. No loops allowed! |
| **Sorting** | `df.sort_values(by='A')` | Ranking the winners. Use `ascending=False` for high-to-low. |
| **Boolean Mask** | `mask = df['A'] > 100` | Asking a True/False question to every row. |
| **Filtering** | `df[mask]` | The Bouncer. Keeps the `True` rows, kicks out the `False` rows. |
